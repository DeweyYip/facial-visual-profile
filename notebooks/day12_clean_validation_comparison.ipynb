```python
from pathlib import Path
import hashlib
import json
import os
import shutil
import sqlite3
import sys
import zipfile

import numpy as np
import torch
from torch.utils.data import DataLoader
from tqdm.auto import tqdm

drive_root = Path("/content/drive/MyDrive/FacialVisualProfile")
r50_run = drive_root / "runs/r50_seed0"
workspace = Path("/content/day12_r50_workspace")
data_root = Path("/content/day12_data")
raw_root = data_root / "raw"
image_root = raw_root / "img_align_celeba"
output_dir = r50_run / "day12"

assert torch.cuda.is_available(), "Select a GPU runtime first."

config = json.loads(
    (r50_run / "config.json").read_text(encoding="utf-8")
)
best = json.loads(
    (r50_run / "best.json").read_text(encoding="utf-8")
)

# Restore the exact source snapshot used for R50 training.
for name, expected_hash in config["source_sha256"].items():
    source = r50_run / "source" / name
    content = source.read_bytes()
    assert hashlib.sha256(content).hexdigest() == expected_hash

    destination = workspace / name
    destination.parent.mkdir(parents=True, exist_ok=True)
    destination.write_bytes(content)

for package in ("datasets", "models", "evaluation"):
    (workspace / package / "__init__.py").touch()

# Prevent accidental reuse of modules imported from another project.
for name, module in list(sys.modules.items()):
    if name.split(".")[0] in {"datasets", "models", "evaluation"}:
        module_path = getattr(module, "__file__", None)
        assert module_path is not None and Path(
            module_path
        ).resolve().is_relative_to(workspace.resolve()), (
            f"Conflicting imported module: {name}. "
            "Restart the runtime and rerun the Day 12 cells."
        )

sys.path.insert(0, str(workspace))
os.chdir(workspace)

from datasets.celeba import CelebAAttributes
from models.resnet50 import ResNet50Attributes
from evaluation.attributes import evaluate_attributes

raw_root.mkdir(parents=True, exist_ok=True)
image_root.mkdir(parents=True, exist_ok=True)

for name in (
    "list_attr_celeba.txt",
    "list_landmarks_align_celeba.txt",
    "list_eval_partition.txt",
):
    shutil.copy2(drive_root / "data" / name, raw_root / name)

valid_names = [
    name
    for line in (
        raw_root / "list_eval_partition.txt"
    ).read_text(encoding="utf-8-sig").splitlines()
    for name, split in [line.split()]
    if split == "1"
]
assert len(valid_names) == 19_867
assert len(set(valid_names)) == len(valid_names)

# Copy the cache locally for image-hash verification.
local_cache = data_root / "mtcnn_clean.sqlite"
shutil.copy2(
    drive_root / "cache/mtcnn_clean.sqlite",
    local_cache,
)
with sqlite3.connect(
    local_cache.resolve().as_uri() + "?mode=ro",
    uri=True,
) as connection:
    cache_records = {
        name: json.loads(result)
        for name, result in connection.execute(
            "SELECT name, result FROM detections WHERE split = ?",
            ("1",),
        )
    }
assert set(cache_records) == set(valid_names)

local_archive = data_root / "img_align_celeba.zip"
if not local_archive.exists():
    print("Copying image archive locally...", flush=True)
    shutil.copy2(
        drive_root / "data/img_align_celeba.zip",
        local_archive,
    )

valid_set = set(valid_names)
with zipfile.ZipFile(local_archive) as archive:
    members = {}
    for member in archive.infolist():
        name = Path(member.filename).name
        if not member.is_dir() and name in valid_set:
            assert name not in members, f"Duplicate ZIP image: {name}"
            members[name] = member

    assert set(members) == valid_set

    for name in tqdm(valid_names, desc="Prepare validation images"):
        content = archive.read(members[name])
        assert hashlib.sha256(content).hexdigest() == (
            cache_records[name]["image_sha256"]
        ), f"Image hash mismatch: {name}"
        (image_root / name).write_bytes(content)

dataset = CelebAAttributes(split="valid", root=raw_root)
assert dataset.names == valid_names
assert dataset.attribute_names == config["attribute_names"]

loader = DataLoader(
    dataset,
    batch_size=config["batch_size"],
    shuffle=False,
    num_workers=0,
    pin_memory=True,
    drop_last=False,
)

checkpoint = torch.load(
    r50_run / best["checkpoint"],
    map_location="cpu",
    weights_only=False,
)
assert checkpoint["epoch"] == best["epoch"]
assert checkpoint["config"] == config

device = torch.device("cuda")
model = ResNet50Attributes(
    num_attributes=len(config["attribute_names"]),
    pretrained=False,
)
model.load_state_dict(checkpoint["model_state"], strict=True)
model.to(device).eval()
del checkpoint

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False

criterion = torch.nn.BCEWithLogitsLoss()
logit_batches = []
target_batches = []
filenames = []
loss_sum = 0.0
sample_count = 0

with torch.inference_mode():
    for images, labels, _, names in tqdm(
        loader, desc="R50 best-checkpoint validation"
    ):
        target_batches.append(labels.numpy().copy())
        filenames.extend(names)

        images = images.to(device, non_blocking=True)
        labels = labels.to(device, non_blocking=True)
        logits = model(images)
        loss = criterion(logits, labels)

        assert torch.isfinite(logits).all()
        assert torch.isfinite(loss)

        count = images.shape[0]
        sample_count += count
        loss_sum += loss.item() * count
        logit_batches.append(logits.cpu().numpy())

logits_array = np.concatenate(logit_batches)
targets_array = np.concatenate(target_batches)
assert sample_count == 19_867
assert filenames == valid_names

# Confirm exact sample, label, and attribute alignment with M3.
with np.load(
    drive_root / "runs/m3_seed0/validation_predictions.npz",
    allow_pickle=False,
) as m3:
    assert np.array_equal(np.asarray(filenames), m3["filenames"])
    assert np.array_equal(targets_array, m3["targets"])
    assert np.array_equal(
        np.asarray(dataset.attribute_names),
        m3["attribute_names"],
    )

metrics = evaluate_attributes(
    logits_array,
    targets_array,
    threshold=config["f1_threshold"],
)
recomputed = {
    "validation_bce": loss_sum / sample_count,
    "validation_macro_f1": float(metrics["macro_f1"]),
    "validation_mAP": float(metrics["map"]),
}
expected = {
    "validation_bce": float(best["valid_loss"]),
    "validation_macro_f1": float(best["validation"]["macro_f1"]),
    "validation_mAP": float(best["validation"]["map"]),
}
differences = {
    key: abs(recomputed[key] - expected[key])
    for key in expected
}
tolerance = 1e-6
passed = all(value <= tolerance for value in differences.values())

output_dir.mkdir(parents=True, exist_ok=True)
prediction_path = output_dir / "validation_predictions.npz"
temporary_prediction = output_dir / "validation_predictions.npz.tmp"

with temporary_prediction.open("wb") as file:
    np.savez_compressed(
        file,
        logits=logits_array,
        targets=targets_array,
        filenames=np.asarray(filenames),
        attribute_names=np.asarray(dataset.attribute_names),
    )
temporary_prediction.replace(prediction_path)

report = {
    "model": "ResNet50Attributes",
    "checkpoint": best["checkpoint"],
    "epoch": best["epoch"],
    "source_commit": config["git_commit"],
    "saved_source_hashes_verified": True,
    "validation_image_hashes_verified": sample_count,
    "sample_label_attribute_alignment_with_m3": True,
    "validation_images": sample_count,
    "f1_threshold": config["f1_threshold"],
    "recomputed": recomputed,
    "expected": expected,
    "absolute_differences": differences,
    "tolerance": tolerance,
    "passed": passed,
    "num_workers": 0,
    "environment": {
        "torch": str(torch.__version__),
        "gpu": torch.cuda.get_device_name(0),
    },
    "test_evaluated": False,
}

report_path = output_dir / "best_verification.json"
temporary_report = output_dir / "best_verification.json.tmp"
temporary_report.write_text(
    json.dumps(report, indent=2, allow_nan=False) + "\n",
    encoding="utf-8",
)
temporary_report.replace(report_path)

print(json.dumps(report, indent=2))
print("Saved:", prediction_path)
print("Saved:", report_path)

assert passed, "Metric mismatch: inspect the report before proceeding."
print("Day 12 R50 best-checkpoint verification: PASS")
```

```python
from pathlib import Path
import json
import math

import torch

drive_root = Path("/content/drive/MyDrive/FacialVisualProfile")
m3_run = drive_root / "runs/m3_seed0"
report_dir = drive_root / "runs/comparisons/day12"

config = json.loads(
    (m3_run / "config.json").read_text(encoding="utf-8")
)
best = json.loads(
    (m3_run / "best.json").read_text(encoding="utf-8")
)

initial_checkpoint = torch.load(
    m3_run / "epoch_00.pt",
    map_location="cpu",
    weights_only=False,
)
trained_checkpoint = torch.load(
    m3_run / best["checkpoint"],
    map_location="cpu",
    weights_only=False,
)

assert initial_checkpoint["epoch"] == 0
assert trained_checkpoint["epoch"] == best["epoch"]
assert initial_checkpoint["config"] == config
assert trained_checkpoint["config"] == config

initial_state = initial_checkpoint["model_state_dict"]
trained_state = trained_checkpoint["model_state_dict"]
assert set(initial_state) == set(trained_state)

def summarize_updates(names, column_slice=None):
    records = []
    initial_squared_norm = 0.0
    delta_squared_norm = 0.0

    for name in names:
        before = initial_state[name]
        after = trained_state[name]
        assert before.shape == after.shape

        if column_slice is not None:
            before = before[:, column_slice]
            after = after[:, column_slice]

        before = before.to(torch.float64)
        after = after.to(torch.float64)

        assert torch.isfinite(before).all()
        assert torch.isfinite(after).all()

        delta = after - before
        changed_elements = int(torch.count_nonzero(delta).item())
        initial_squared_norm += float(before.square().sum().item())
        delta_squared_norm += float(delta.square().sum().item())

        records.append({
            "name": name,
            "elements": before.numel(),
            "changed_elements": changed_elements,
            "max_absolute_change": float(delta.abs().max().item()),
        })

    initial_norm = math.sqrt(initial_squared_norm)
    delta_norm = math.sqrt(delta_squared_norm)

    return {
        "parameter_tensors": len(records),
        "changed_parameter_tensors": sum(
            record["changed_elements"] > 0 for record in records
        ),
        "parameter_elements": sum(
            record["elements"] for record in records
        ),
        "changed_parameter_elements": sum(
            record["changed_elements"] for record in records
        ),
        "delta_l2": delta_norm,
        "relative_delta_l2": (
            delta_norm / initial_norm if initial_norm > 0 else None
        ),
        "parameters": records,
    }

# Select learned weights and biases, excluding BatchNorm running buffers.
groups = {}
for prefix in ("image_encoder", "geometry_encoder", "classifier"):
    names = sorted(
        name for name in initial_state
        if name.startswith(prefix + ".")
        and name.endswith((".weight", ".bias"))
    )
    assert names, f"No parameters found for {prefix}"
    groups[prefix] = summarize_updates(names)

assert initial_state["classifier.weight"].shape == (24, 544)

# Check classifier connections to each feature branch separately.
groups["classifier_image_columns"] = summarize_updates(
    ["classifier.weight"],
    column_slice=slice(0, 512),
)
groups["classifier_geometry_columns"] = summarize_updates(
    ["classifier.weight"],
    column_slice=slice(512, 544),
)

passed = all(
    group["changed_parameter_elements"] > 0
    for group in groups.values()
)

report = {
    "source_commit": config["git_commit"],
    "initial_checkpoint": "epoch_00.pt",
    "trained_checkpoint": best["checkpoint"],
    "trained_epoch": best["epoch"],
    "method": "Compare learned weights and biases; exclude running buffers.",
    "groups": groups,
    "passed": passed,
    "interpretation": (
        "Parameter changes confirm training updates in both branches "
        "and classifier connections. They do not establish a causal "
        "performance benefit from geometry."
    ),
    "optimizer_updates_performed_by_this_check": False,
    "test_evaluated": False,
}

report_dir.mkdir(parents=True, exist_ok=True)
report_path = report_dir / "m3_branch_update_report.json"
temporary = report_path.with_suffix(".json.tmp")
temporary.write_text(
    json.dumps(report, indent=2, allow_nan=False) + "\n",
    encoding="utf-8",
)
temporary.replace(report_path)

for name, group in groups.items():
    print(
        f"{name}: "
        f"changed tensors="
        f"{group['changed_parameter_tensors']}/"
        f"{group['parameter_tensors']}, "
        f"changed elements={group['changed_parameter_elements']}, "
        f"delta L2={group['delta_l2']:.8f}"
    )

print("Saved:", report_path)
assert passed, "A parameter group has no updates; inspect the report."
print("Day 12 M3 branch parameter-update check: PASS")
```

```python
from pathlib import Path
import hashlib
import json

import numpy as np
import pandas as pd

from evaluation.attributes import evaluate_attributes

drive_root = Path("/content/drive/MyDrive/FacialVisualProfile")
runs_root = drive_root / "runs"
report_dir = runs_root / "comparisons/day12"
report_dir.mkdir(parents=True, exist_ok=True)

prediction_paths = {
    "R18": runs_root / "r18_seed0/day7/validation_predictions.npz",
    "R50": runs_root / "r50_seed0/day12/validation_predictions.npz",
    "M3": runs_root / "m3_seed0/validation_predictions.npz",
}
run_names = {
    "R18": "r18_seed0",
    "R50": "r50_seed0",
    "M3": "m3_seed0",
}

# Require successful fresh-checkpoint verification for R50 and M3.
for path in (
    runs_root / "r50_seed0/day12/best_verification.json",
    runs_root / "m3_seed0/best_verification.json",
):
    assert json.loads(path.read_text(encoding="utf-8"))["passed"] is True

arrays = {}
configs = {}
best_records = {}

for model_name, path in prediction_paths.items():
    with np.load(path, allow_pickle=False) as archive:
        arrays[model_name] = {
            key: archive[key].copy()
            for key in (
                "logits", "targets", "filenames", "attribute_names"
            )
        }

    run_path = runs_root / run_names[model_name]
    configs[model_name] = json.loads(
        (run_path / "config.json").read_text(encoding="utf-8")
    )
    best_records[model_name] = json.loads(
        (run_path / "best.json").read_text(encoding="utf-8")
    )

reference = arrays["R18"]
attribute_names = reference["attribute_names"].tolist()

partition = (
    drive_root / "data/list_eval_partition.txt"
).read_text(encoding="utf-8-sig")
official_valid_names = [
    name
    for line in partition.splitlines()
    for name, split in [line.split()]
    if split == "1"
]
assert reference["filenames"].tolist() == official_valid_names
assert len(official_valid_names) == 19_867
assert len(attribute_names) == 24

shared_fields = (
    "seed", "epochs", "batch_size", "optimizer",
    "learning_rate", "weight_decay", "precision",
    "augmentation", "f1_threshold", "selection_metric",
    "epoch_seed_rule", "cudnn_deterministic", "cudnn_benchmark",
)

for model_name, data in arrays.items():
    assert data["logits"].shape == (19_867, 24)
    assert data["targets"].shape == (19_867, 24)
    assert np.isfinite(data["logits"]).all()
    assert np.isin(data["targets"], [0, 1]).all()

    for key in ("targets", "filenames", "attribute_names"):
        assert np.array_equal(data[key], reference[key]), (
            f"{model_name}: alignment mismatch for {key}"
        )

    config = configs[model_name]
    assert config["attribute_names"] == attribute_names
    assert config["f1_threshold"] == 0.5

    for field in shared_fields:
        assert field in config and field in configs["R18"]
        assert config[field] == configs["R18"][field], (
            f"{model_name}: protocol mismatch for {field}"
        )

def expected_metrics(model_name):
    best = best_records[model_name]
    if model_name == "R18":
        return best["validation_map"], best["validation_macro_f1"]
    if model_name == "R50":
        return best["validation"]["map"], best["validation"]["macro_f1"]
    return best["validation_mAP"], best["validation_macro_f1"]

summary_rows = []
per_attribute = pd.DataFrame({
    "attribute": attribute_names,
    "positive_count": reference["targets"].sum(axis=0).astype(int),
})
all_metrics = {}

for model_name, data in arrays.items():
    metrics = evaluate_attributes(
        data["logits"], data["targets"], threshold=0.5
    )
    expected_map, expected_f1 = expected_metrics(model_name)
    assert abs(float(metrics["map"]) - expected_map) <= 1e-6
    assert abs(float(metrics["macro_f1"]) - expected_f1) <= 1e-6

    # Stable BCE from logits, computed uniformly for all models.
    logits = data["logits"].astype(np.float64)
    targets = data["targets"].astype(np.float64)
    bce = float(
        np.mean(np.logaddexp(0.0, logits) - targets * logits)
    )

    summary_rows.append({
        "model": model_name,
        "best_epoch": best_records[model_name]["epoch"],
        "validation_images": len(data["filenames"]),
        "BCE": bce,
        "macro_F1": float(metrics["macro_f1"]),
        "mAP": float(metrics["map"]),
        "F1_threshold": 0.5,
    })
    per_attribute[f"{model_name}_AP"] = metrics["per_attribute_ap"]
    per_attribute[f"{model_name}_F1"] = metrics["per_attribute_f1"]
    all_metrics[model_name] = {
        "mAP": float(metrics["map"]),
        "macro_F1": float(metrics["macro_f1"]),
        "BCE": bce,
    }

# Differences are percentage points, not relative percentages.
for baseline in ("R18", "R50"):
    per_attribute[f"M3_minus_{baseline}_AP_pp"] = 100 * (
        per_attribute["M3_AP"] - per_attribute[f"{baseline}_AP"]
    )
    per_attribute[f"M3_minus_{baseline}_F1_pp"] = 100 * (
        per_attribute["M3_F1"] - per_attribute[f"{baseline}_F1"]
    )

summary = pd.DataFrame(summary_rows)
summary.to_csv(report_dir / "clean_validation_summary.csv", index=False)
per_attribute.to_csv(
    report_dir / "clean_validation_per_attribute.csv", index=False
)

comparison = {
    "split": "official validation",
    "validation_images": 19_867,
    "attributes": attribute_names,
    "sample_label_attribute_alignment": True,
    "shared_training_protocol_checked": list(shared_fields),
    "f1_threshold": 0.5,
    "metrics": all_metrics,
    "prediction_files": {
        name: {
            "path": str(path.relative_to(drive_root)),
            "sha256": hashlib.sha256(path.read_bytes()).hexdigest(),
        }
        for name, path in prediction_paths.items()
    },
    "source_commits": {
        name: config["git_commit"]
        for name, config in configs.items()
    },
    "scope": (
        "Single-seed clean-validation comparison of checkpoints selected "
        "by validation mAP. No significance or causal benefit established."
    ),
    "test_evaluated": False,
}
(report_dir / "clean_validation_comparison.json").write_text(
    json.dumps(comparison, indent=2, allow_nan=False) + "\n",
    encoding="utf-8",
)

print("=== Overall clean-validation comparison ===")
print(summary.to_string(index=False, float_format=lambda x: f"{x:.6f}"))

print("\n=== Per-attribute AP comparison ===")
print(per_attribute[[
    "attribute", "R18_AP", "R50_AP", "M3_AP",
    "M3_minus_R18_AP_pp", "M3_minus_R50_AP_pp",
]].to_string(index=False, float_format=lambda x: f"{x:.4f}"))

print("\n=== Per-attribute F1 comparison ===")
print(per_attribute[[
    "attribute", "R18_F1", "R50_F1", "M3_F1",
    "M3_minus_R18_F1_pp", "M3_minus_R50_F1_pp",
]].to_string(index=False, float_format=lambda x: f"{x:.4f}"))

print("\nSaved reports:", report_dir)
print("Day 12 clean-validation comparison: PASS")
```
